## 1. Setup & Data Loading

```python
# Install dependencies if needed (uncomment for Google Colab)
# !pip install pandas numpy matplotlib seaborn scipy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='Set2')
plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.size'] = 12

print('✅ Libraries loaded successfully')
```

```python
 !ls Data/LLSP/annotation
```

```python
# If running on Google Colab, upload the file first:
# from google.colab import files
# uploaded = files.upload()

CSV_PATH = 'Data/LLSP/annotation/train.csv'   # ← change path if needed

df = pd.read_csv(CSV_PATH)

if df.columns[0] in ['Unnamed: 0', '']:
    df.rename(columns={df.columns[0]: 'index'}, inplace=True)

meta_cols = ['index', 'type', 'name', 'count']
L_cols    = [c for c in df.columns if c.startswith('L')]

print(f'Shape          : {df.shape}')
print(f'Meta columns   : {meta_cols}')
print(f'Location cols  : {len(L_cols)}  ({L_cols[0]} … {L_cols[-1]})')
df.head()
```

## 2. Dataset Overview

```python
print('=' * 55)
print('GENERAL DATASET OVERVIEW')
print('=' * 55)
print(f"Total videos         : {len(df)}")
print(f"Unique action types  : {df['type'].nunique()}")
print(f"Total columns        : {df.shape[1]}")
print(f"Location columns (L) : {len(L_cols)}")
print()
print('── Data Types ──')
print(df[meta_cols].dtypes)
print()
print('── Count column stats ──')
print(df['count'].describe().round(2))
```

```python
print('Sample videos per action type:')
df.groupby('type')[['name','count']].head(2).reset_index()
```

## 3. Class Imbalance Analysis

```python
class_counts = df['type'].value_counts().reset_index()
class_counts.columns = ['type', 'count_videos']
class_counts['percentage']   = (class_counts['count_videos'] / len(df) * 100).round(2)
class_counts['cumulative_%'] = class_counts['percentage'].cumsum().round(2)

print(class_counts.to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

colors = sns.color_palette('Set2', len(class_counts))
bars = axes[0].barh(class_counts['type'], class_counts['count_videos'],
                    color=colors, edgecolor='white')
axes[0].set_xlabel('Number of Videos')
axes[0].set_title('Video Count per Action Type', fontweight='bold')
axes[0].invert_yaxis()
for bar, val in zip(bars, class_counts['count_videos']):
    axes[0].text(bar.get_width() + 1, bar.get_y() + bar.get_height() / 2,
                 f'{val}', va='center', fontsize=9)

wedges, texts, autotexts = axes[1].pie(
    class_counts['count_videos'],
    labels=class_counts['type'],
    autopct='%1.1f%%',
    colors=colors,
    startangle=140,
    pctdistance=0.82
)
for t in autotexts:
    t.set_fontsize(8)
axes[1].set_title('Class Distribution (%)', fontweight='bold')

plt.suptitle('RepCount Train Set — Class Imbalance Analysis', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('class_imbalance.png', bbox_inches='tight')
plt.show()

imbalance_ratio = class_counts['count_videos'].max() / class_counts['count_videos'].min()
print(f'\n⚠️  Imbalance Ratio (max/min): {imbalance_ratio:.2f}x')
```

## 4. Repetition Count Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].hist(df['count'], bins=30, color=sns.color_palette('Set2')[0],
             edgecolor='white', linewidth=0.8)
axes[0].axvline(df['count'].mean(),   color='red',    linestyle='--', label=f"Mean = {df['count'].mean():.1f}")
axes[0].axvline(df['count'].median(), color='orange', linestyle='--', label=f"Median = {df['count'].median():.1f}")
axes[0].set_xlabel('Number of Repetitions')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Distribution of Repetition Counts (all classes)', fontweight='bold')
axes[0].legend()

order = df.groupby('type')['count'].median().sort_values(ascending=False).index
sns.boxplot(data=df, x='count', y='type', order=order, palette='Set2', ax=axes[1])
axes[1].set_xlabel('Number of Repetitions')
axes[1].set_ylabel('Action Type')
axes[1].set_title('Repetition Count Distribution per Class', fontweight='bold')

plt.suptitle('Repetition Count Analysis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('count_distribution.png', bbox_inches='tight')
plt.show()
```

```python
count_stats = df.groupby('type')['count'].agg(['mean','median','std','min','max']).round(2)
count_stats.columns = ['Mean', 'Median', 'Std', 'Min', 'Max']
count_stats = count_stats.sort_values('Mean', ascending=False)
print('Per-Class Repetition Count Statistics:')
print(count_stats.to_string())
```

## 5. Temporal Annotation Analysis (L columns)

```python
df['active_L_cols']    = df[L_cols].notna().sum(axis=1)
df['annotated_cycles'] = (df['active_L_cols'] / 2).round(0)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].hist(df['active_L_cols'], bins=30,
             color=sns.color_palette('Set2')[1], edgecolor='white')
axes[0].set_xlabel('Number of Non-NaN L columns')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Active Annotation Columns per Video', fontweight='bold')

axes[1].scatter(df['count'], df['annotated_cycles'],
                alpha=0.5, color=sns.color_palette('Set2')[2], edgecolors='none')
lims = [0, max(df['count'].max(), df['annotated_cycles'].max()) + 2]
axes[1].plot(lims, lims, 'r--', linewidth=1.5, label='Perfect agreement')
axes[1].set_xlabel('count column (ground truth)')
axes[1].set_ylabel('Annotated cycles (from L cols)')
axes[1].set_title('count vs. Temporal Annotations Consistency', fontweight='bold')
axes[1].legend()

plt.suptitle('Temporal Annotation (L Columns) Analysis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('temporal_annotation.png', bbox_inches='tight')
plt.show()

mismatch = df[df['count'] != df['annotated_cycles']]
print(f'Total videos             : {len(df)}')
print(f'count == annotated_cycles: {len(df) - len(mismatch)}')
print(f'Mismatches               : {len(mismatch)}  ({100*len(mismatch)/len(df):.1f}%)')
```

```python
fig, ax = plt.subplots(figsize=(16, 6))
sample = df[L_cols[:60]].isna()
sns.heatmap(sample, cmap='Blues_r', cbar=False, yticklabels=False, ax=ax,
            xticklabels=[c if i % 5 == 0 else '' for i, c in enumerate(L_cols[:60])])
ax.set_title('NaN Sparsity of First 60 L-columns\n(white = annotated, blue = NaN)', fontweight='bold')
ax.set_xlabel('L column')
ax.set_ylabel('Video (row)')
plt.tight_layout()
plt.savefig('nan_sparsity.png', bbox_inches='tight')
plt.show()
```

## 6. Missing Data Analysis

```python
print('Missing values in metadata columns:')
print(df[meta_cols].isnull().sum())

total_L_cells   = df[L_cols].size
missing_L_cells = df[L_cols].isnull().sum().sum()
print(f'\nL columns — total cells : {total_L_cells:,}')
print(f'L columns — NaN cells   : {missing_L_cells:,}  ({100*missing_L_cells/total_L_cells:.1f}%)')
print(f'L columns — filled cells: {total_L_cells - missing_L_cells:,}  ({100*(1-missing_L_cells/total_L_cells):.1f}%)')
```

```python
print('Max annotation depth (highest L index used) per class:')
for cls in df['type'].unique():
    sub    = df[df['type'] == cls][L_cols]
    last_l = sub.notna().any(axis=0)
    last_idx = last_l[last_l].index[-1] if last_l.any() else 'N/A'
    print(f'  {cls:<25} → max col used: {last_idx}')
```

## 7. Per-Class Deep-Dive

```python
fig, ax = plt.subplots(figsize=(14, 6))
order = df.groupby('type')['count'].median().sort_values(ascending=False).index
sns.violinplot(data=df, x='type', y='count', order=order,
               palette='Set2', inner='quartile', ax=ax)
ax.set_xlabel('Action Type')
ax.set_ylabel('Repetition Count')
ax.set_title('Repetition Count Distribution per Class (Violin)', fontweight='bold')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('violin_per_class.png', bbox_inches='tight')
plt.show()
```

```python
threshold = df['count'].mean() + 3 * df['count'].std()
outliers  = df[df['count'] > threshold][['type', 'name', 'count']].sort_values('count', ascending=False)
print(f'Outlier threshold (mean + 3σ): {threshold:.1f} repetitions')
print(f'Number of outlier videos     : {len(outliers)}')
print()
print(outliers.to_string(index=False))
```

```python
def avg_cycle_duration(row):
    vals = [row[c] for c in L_cols if not pd.isna(row[c])]
    if len(vals) < 2:
        return np.nan
    durations = [vals[i+1] - vals[i] for i in range(0, len(vals)-1, 2)]
    return np.mean(durations) if durations else np.nan

df['avg_cycle_duration'] = df.apply(avg_cycle_duration, axis=1)

fig, ax = plt.subplots(figsize=(14, 5))
order2 = df.groupby('type')['avg_cycle_duration'].median().sort_values(ascending=False).index
sns.barplot(data=df, x='type', y='avg_cycle_duration', order=order2,
            palette='Set2', ci='sd', ax=ax)
ax.set_xlabel('Action Type')
ax.set_ylabel('Avg Cycle Duration (frames)')
ax.set_title('Average Repetition Cycle Duration per Class (±1 SD)', fontweight='bold')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig('cycle_duration.png', bbox_inches='tight')
plt.show()
```
