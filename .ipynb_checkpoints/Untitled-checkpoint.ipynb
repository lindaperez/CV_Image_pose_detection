```python
# Check if the video exist 
videos_path = "Data/LLSP/video/train/"
import cv2

def play_video(video_path):
    # Create a video capture object
    cap = cv2.VideoCapture(video_path)

    # Check if the video file was opened successfully
    if not cap.isOpened():
        print("Error: Could not open video file.")
        return

    # Loop through the frames of the video
    while True:
        ret, frame = cap.read() # Read a frame

        # If the frame was not read correctly, the video has ended
        if not ret:
            break

        # Display the frame
        cv2.imshow('Video Playback (Press "q" to quit)', frame)

        # Press 'q' on the keyboard to exit the playback
        # waitKey(25) determines the speed of playback
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break

    # Release the video capture object and close all windows
    cap.release()
    cv2.destroyAllWindows()


```

```python
# Replace 'your_video_file.mp4' with the path to your video file
#play_video(str(squat['name'][11]))
#play_video('stu9_49.mp4')
# Replace 'your_video.mp4' with the actual path to your MP4 file
play_video(videos_path+'stu9_49.mp4')
```
